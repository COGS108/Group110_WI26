## Research Question

Did Bitcoin behave more like a safe-haven asset or risk asset before, during, and after the COVID-19 pandemic (2017-2025), based on its correlations with the S&P 500, Gold price index, and U.S. Dollar index?

## Background and Prior Work

Over the past decade, cryptocurrencies–especially Bitcoin—have transitioned from niche digital assets into instruments increasingly integrated into global financial markets. As governmental and institutional participation has grown, researchers and investors have become more interested in understanding how Bitcoin interacts and compares with traditional macro assets. Namely: equities, commodities and currencies are subject to being juxtaposed with Bitcoin in consideration of their comparative investment advantages during such an unbridled economic period such as the Covid-19 era. 
We define the “Covid-19 era” to take place from the year 2017 through 2025, for our purposes. Specifically, the “pre-Covid era” will be the three years before the pandemic’s inception: 2017-2019; the “Covid era” will be the three years following its inception including its peak: 2020-2022; and the “post-Covid era” will be the three years following its peak: 2023-2025. Understanding whether Bitcoin behaves more like a speculative technology stock, a safe haven such as gold, or a liquidity sensitive macro asset is important for portfolio construction, risk management, and macroeconomic analysis.

Prior research suggests Bitcoin price historically exhibited low correlations with traditional assets before 2020. For example, Baur et al., analyzed Bitcoin's statistical properties and role as a diversification asset and found weak correlations with stocks and gold, concluding that bitcoin behaved more like a speculative investment than a safe haven or hedging tool.<sup><a href="#ref1">1</a></sup> Similarly, Corbet et al. examined spillovers between cryptocurrencies and financial markets and reported limited integration between crypto and traditional assets during Bitcoin's early growth phase.<sup><a href="#ref2">2</a></sup> These findings substantiate the early narrative that Bitcoin could serve as a portfolio diversifier due to its independence from typical macro markets.

However, more recent studies suggest that the Covid-19 pandemic marked a structural shift in asset correlations. Conlon and McGee examined Bitcoin’s performance during the March 2020 market crash and found it failed to act as a safe haven. Instead, Bitcoin exhibited increasing co-movement in price with equites during extreme volatility.<sup><a href="#ref3">3</a></sup> In a broader time-varying analysis, Dutta et al. documented dynamic correlations between gold, and stock markets, finding that Bitcoin’s relationship with equities strengthened during periods of market stress.<sup><a href="#ref4">4</a></sup> These findings suggest that correlations involving Bitcoin are not static but depend heavily on macroeconomic regimes.

Despite this growing body of literature, relatively few student or open-source data science projects explicitly examine how correlations between Bitcoin returns and multiple macro assets evolve across distinct time regimes, such as pre-Covid, Covid, and post-Covid periods; at least that we were able to find. Our project builds on existing literature by performing a regime-based correlation analysis of daily percentage returns across several major asset classes–the S&P 500 (stocks), gold (commodity), and the U.S. dollar index (currency)–over the years of 2017-2025. By combining rolling correlations and period segmentation, this work aims to evaluate whether Bitcoin has structurally transitioned from an uncorrelated, technological alternative currency into a macro-sensitive risk asset.

## Hypothesis


We hypothesize that: from 2017 to 2025 in the United States, the relationship between daily percentage returns of Bitcoin and traditional macro-financial assets—the S&P 500 Index, gold price, and the U.S. Dollar Index (DXY)—changed significantly across the pre-COVID (2017–2019), COVID (2020–2022), and post-COVID (2023–2025) periods. 

Prior to the COVID-19 pandemic, Bitcoin exhibited weak or negligible correlations with equities, gold, and the U.S. dollar, consistent with its characterization as an alternative or speculative asset; according to the paper by Baur et al.

During the COVID-19 period, Bitcoin’s correlation with the S&P 500 increased significantly while its correlations with gold and the dollar failed to strengthen in a manner consistent with safe-haven behavior, reflecting heightened integration with risk-on financial markets (i.e. stocks) during periods of systemic stress. 

In the post-COVID period, these elevated equity correlations persisted or stabilized, suggesting that Bitcoin has structurally transitioned toward behaving as a macro-sensitive risk asset rather than an independent currency. 

This hypothesis is tested by examining changes in correlation structure and regression coefficients across the three time regimes using daily price data that is statistically transformed into daily percentage returns; guided by the corpus of literature and previous scientific and research studies conducted in our referenced papers.

### Dataset #4
#### US Dollar Index Prices

Within this dataset there are six columns. They are date, close, high, low, open, and volume. The most important metric that will be used for the project is the close column. The Close column represents the final reported value of the U.S. Dollar Index at the end of each trading session. It is measured in index points. These are unitless values that represent the relative strength of the U.S dollar compared to a basket of major foreign currencies. Those being: the Euro (EUR), Japanese Yen (JPY), British Pound (GBP), Canadian Dollar (CAD), Swedish Krona (SEK), and Swiss Franc (CHF).

The close is the most important daily metric because it serves as the official benchmark for performance measurement. Daily returns are calculated using changes in closing prices. Technical indicators such as the Relative Strength Index (RSI) are also calculated using this metric, which is a tool used to measure the speed and change of price movements. Moreover, the close tells us whether the dollar strengthend or weakened in relative to a previous day. For example, if on some day the dollar closes at 80 and the next day the dollar closes at 90, this means the dollar has strengthened, whereas if the dollar first closed at 90 and then closed at 80 this would mean the dollar has weakened.

For the past few decades, the USD index has fluctuated between 80 and 115. Values above 110 generally reflect a strong dollar, whereas values below 85 suggest a weak dollar.

In short, the close column for the USD Index lets us measure the relative strength and weakness of the US Dollar over a period of time.

There are no major concerns with the data. The data was taken from Yahoo Finance, which is a trusted source for market statistics.

**Important note:** The number of support observations or row entries is practially the same for all DataFrames except `df_btc`. This is due to the aformentioned reason that---unlike tradtional financial markets which track stocks, currencies, and commodities---cryptocurrency markets do not close and then open for regular trading sessions, they are instead open 24 hours a day, 7 days a week; leading to the 1.5x increase in support observations or trading days.

```python
df_btc['BTC Close']= pd.to_numeric(df_btc['BTC Close'], errors="coerce")
df_snp['S&P Close']= pd.to_numeric(df_snp['S&P Close'], errors="coerce")
df_gold['GOLD Close']= pd.to_numeric(df_gold['GOLD Close'], errors="coerce")
df_usd['USD Close']= pd.to_numeric(df_usd['USD Close'], errors="coerce")

print(df_btc.dtypes)
print(df_snp.dtypes)
print(df_gold.dtypes)
print(df_usd.dtypes)
```

```python
# Dropping unecessary columns ('High','Low','Open','Volume') which are duplicates 
# across all 4 tables before inner joining
df_btc = df_btc[['Date', 'BTC Close']]
df_snp = df_snp[['Date', 'S&P Close']]
df_gold = df_gold[['Date', 'GOLD Close']]
df_usd = df_usd[['Date', 'USD Close']]

# Using the `log` and `diff` methods to create a column of daily returns instead of daily price
df_btc["btc_log_return"] = np.log(df_btc["BTC Close"]).diff()
df_snp["s&p_log_return"] = np.log(df_snp["S&P Close"]).diff()
df_gold["gold_log_return"] = np.log(df_gold["GOLD Close"]).diff()
df_usd["usd_log_return"] = np.log(df_usd["USD Close"]).diff()

# Since the very first row won't have a previous day of closing price to use as a reference in calculating
# a percentage return, the `log` and `diff` methods will make it `Nan`. 
# So we are manually changing the `NaNs` is the first row to `0` here
df_btc.fillna(0, inplace=True)
df_snp.fillna(0, inplace=True)
df_gold.fillna(0, inplace=True)
df_usd.fillna(0, inplace=True)

# Now dropping the daily closing price columns to only be left with a column for 'Date'
# and a column for the log daily percent return; a more "continous" percentage than not log transforming returns
df_btc = df_btc[['Date', 'btc_log_return']]
df_snp = df_snp[['Date', 's&p_log_return']]
df_gold = df_gold[['Date', 'gold_log_return']]
df_usd = df_usd[['Date', 'usd_log_return']]

# inner joining tables using the Date column as the key
merged_df = df_btc.merge(df_snp, on='Date').merge(df_gold, on='Date').merge(df_usd, on='Date')
merged_df = merged_df.sort_values('Date')
merged_df
```

```python
# dataframes representing a time period related to the COVID-19 pandemic, 
# specifically pre-, during-, and post-COVID-19

df_pre = merged_df.loc[0:799].reset_index(drop=True) # Get pre-pandemic from Jan 2017 to March 10th, 2020 
df_pandemic = merged_df.loc[800:1593].reset_index(drop=True) # Covid was declared a pandemic on March 11th, 2020 by the WHO
df_post = merged_df.loc[1594:2260].reset_index(drop=True) # Covid was declared no longer a Public Health Emergency on May 5th, 2023
```

We are generating QQ plots to inspect the normality of the data per financial class variable (btc, s&p, gold, usd), and per Covid-era type (pre-, during-, post-pandemic):

- **Statistical interpretation**: From the test of normality, the resulting p-value of `6.05e-16` suggests a minimal chance of observing this data under the assumption that it is normally distrubuted, and is much smaller than $\alpha=0.05$. This suggests that log-transformed Bitcoin daily returns *before* the Pandemic are not approximately normal.

 - [X] **A.1 Informed consent**: If there are human subjects, have they given informed consent, where subjects affirmatively opt-in and have a clear understanding of the data uses to which they consent?

    > Our data is collected using the open-source tool yfinance from Yahoo! Finance. Yahoo aggregates Bitcoin price data from CoinMarketCap, which use volume-weighted average prices (VWAP), hence high-volume exchanges in certain regions may disproportionately affect the final price. The S&P 500 index has survivorship bias concerns, in which historical performances appear better due to excluding poor-performing firms. In response, we plan to clarify the index data as reflecting the large-cap market rather than generalizing economic well-being during the pandemic.

 - [X] **A.3 Limit PII exposure**: Have we considered ways to minimize exposure of personally identifiable information (PII) for example through anonymization or not collecting information that isn't relevant for analysis?
