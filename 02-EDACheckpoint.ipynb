### Gold Index Daily Prices

This dataset contains daily observations of Gold Futures (GC=F) prices from January 2017 through December 2025. Gold futures are standardized contracts used as a global benchmark for gold price movements. All price values are reported in U.S. dollars (USD) per troy ounce, the standard unit for precious metals trading. The dataset includes six primary variables: Date, Open, High, Low, Close, and Volume.

The Open price represents the first transaction price of the trading session, while the High and Low values capture the maximum and minimum prices observed during the day. The Close price reflects the final traded value and is commonly interpreted as the market’s consensus valuation at the end of trading; for example, a Close price of 1,800 indicates that one troy ounce of gold was valued at $1,800 at market close. Volume measures the number of futures contracts traded and serves as an indicator of market participation and liquidity.

Gold is widely considered a safe-haven asset and an inflation hedge, making it relevant for analyzing financial market behavior during periods of economic uncertainty, including the COVID-19 pandemic. The dataset’s time span allows for comparison of gold price dynamics across pre-pandemic, pandemic, and recovery periods.

Finally, gold prices are influenced by macroeconomic and geopolitical factors not directly captured in this dataset, meaning observed relationships should be interpreted as associative rather than causal.

### US Dollar Index Prices

Within this dataset there are six columns. They are date, close, high, low, open, and volume. The most important metric that will be used for the project is the close column. The Close column represents the final reported value of the U.S. Dollar Index at the end of each trading session. It is measured in index points. These are unitless values that represent the relative strength of the U.S dollar compared to a basket of major foreign currencies. Those being: the Euro (EUR), Japanese Yen (JPY), British Pound (GBP), Canadian Dollar (CAD), Swedish Krona (SEK), and Swiss Franc (CHF).

The close is the most important daily metric because it serves as the official benchmark for performance measurement. Daily returns are calculated using changes in closing prices. Technical indicators such as the Relative Strength Index (RSI) are also calculated using this metric, which is a tool used to measure the speed and change of price movements. Moreover, the close tells us whether the dollar strengthend or weakened in relative to a previous day. For example, if on some day the dollar closes at 80 and the next day the dollar closes at 90, this means the dollar has strengthened, whereas if the dollar first closed at 90 and then closed at 80 this would mean the dollar has weakened.

For the past few decades, the USD index has fluctuated between 80 and 115. Values above 110 generally reflect a strong dollar, whereas values below 85 suggest a weak dollar.

In short, the close column for the USD Index lets us measure the relative strength and weakness of the US Dollar over a period of time.

There are no major concerns with the data. The data was taken from Yahoo Finance, which is a trusted source for market statistics.

### Stationary Tests for the Pre-Covid Pandemic Era 
In order to learn if there exists any correlations between the Bitcoin Returns and the returns of other variables, we must first test each variable for each era for stationarity. If they all are, we can proceed to conduct rolling correlations.

```python
def run_adf_test(series, series_name="Series"):
    """
    Runs Augmented Dickey-Fuller test and prints formatted output.
    """
    result = adfuller(series.dropna())
    
    print(f"\nADF Test Results for {series_name}")
    print("-" * 40)
    print(f"ADF Statistic: {result[0]}")
    print(f"p-value: {result[1]}")
    print("Critical Values:")
    
    for key, value in result[4].items():
        print(f"   {key}: {value}")
    
    if result[1] < 0.05:
        print("Conclusion: Reject H0 → Series is stationary.")
    else:
        print("Conclusion: Fail to reject H0 → Series is non-stationary.")
```
