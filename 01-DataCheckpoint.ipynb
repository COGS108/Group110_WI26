## Gold Futures Price Dataset (GC=F)

### Dataset Description

This dataset contains 2,263 daily observations of Gold Futures (GC=F) prices spanning January 2017 through December 2025. Prices are reported in U.S. dollars (USD) per troy ounce, which is the internationally recognized standard unit for precious metals trading. The dataset includes the variables Open, High, Low, and Close prices, each representing different aspects of daily market activity. The Open price reflects the first traded price of the session, the High and Low values represent the maximum and minimum transaction prices observed during the trading day, and the Close price captures the final market consensus value at the end of trading. For example, a Close price of 1,800 USD indicates that market participants valued one troy ounce of gold at 1,800 USD at the end of that trading session.

The Volume variable measures the number of futures contracts traded during each session and serves as a proxy for market participation and liquidity. Higher trading volume often reflects stronger investor conviction and information flow, while lower volume may indicate reduced market interest or trading activity. Because gold is widely considered a safe-haven asset and an inflation hedge, its price dynamics are frequently analyzed alongside equities, currencies, and cryptocurrencies to understand portfolio diversification and risk sentiment across financial markets.

The dataset spans multiple macroeconomic regimes, including pre-pandemic stability, the COVID-19 pandemic shock, and subsequent recovery periods. This temporal coverage makes the dataset particularly useful for examining whether gold maintained its traditional role as a defensive asset during periods of heightened uncertainty and financial volatility. The long time horizon also supports trend analysis and comparative evaluation with other asset classes included in this project.

### Dataset Concerns and Limitations

Several limitations must be considered when interpreting this dataset. First, the observations represent futures contract prices rather than physical spot gold prices. Futures markets may incorporate speculative behavior, leverage effects, and contract rollover dynamics that can cause deviations from underlying physical market fundamentals. Second, financial market datasets inherently exclude weekends and market holidays, resulting in missing calendar dates that reflect non-trading periods rather than data collection errors.

Additionally, the raw dataset contains formatting artifacts from the Yahoo Finance export process, including metadata rows preceding the actual observations. These rows must be removed during preprocessing to produce a tidy analytical dataset. Missing price entries may occur due to market closures or recording gaps, while extreme price or volume observations may correspond to periods of elevated volatility or macroeconomic stress. Consequently, careful data cleaning and outlier inspection are required prior to statistical analysis.

Finally, gold prices are influenced by numerous external factors, including monetary policy expectations, geopolitical events, inflation dynamics, and currency movements. Because these determinants are not directly observed within the dataset, empirical relationships estimated using this data should be interpreted as associative rather than strictly causal.
