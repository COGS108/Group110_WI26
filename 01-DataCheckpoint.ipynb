## Research Question

Did Bitcoin behave more like a safe-haven asset or risk asset before, during, and after the COVID-19 pandemic (2017-2025), based on its correlations with the S&P 500, Gold price index, and U.S. Dollar index?

## Background and Prior Work

Over the past decade, cryptocurrencies–especially Bitcoin—have transitioned from niche digital assets into instruments increasingly integrated into global financial markets. As governmental and institutional participation has grown, researchers and investors have become more interested in understanding how Bitcoin interacts and compares with traditional macro assets. Namely: equities, commodities and currencies are subject to being juxtaposed with Bitcoin in consideration of their comparative investment advantages during such an unbridled economic period such as the Covid-19 era. 
We define the “Covid-19 era” to take place from the year 2017 through 2025, for our purposes. Specifically, the “pre-Covid era” will be the three years before the pandemic’s inception: 2017-2019; the “during-Covid era” will be the three years following its inception including its peak: 2020-2022; and the “post-Covid era” will be the three years following its peak: 2023-2025. Understanding whether Bitcoin behaves more like a speculative technology stock, a safe haven such as gold, or a liquidity sensitive macro asset is important for portfolio construction, risk management, and macroeconomic analysis.

Prior research suggests Bitcoin price historically exhibited low correlations with traditional assets before 2020. For example, Baur et al., analyzed Bitcoin's statistical properties and role as a diversification asset and found weak correlations with stocks and gold, concluding that bitcoin behaved more like a speculative investment than a safe haven or hedging tool.<sup><a href="#ref1">1</a></sup> Similarly, Corbet et al. examined spillovers between cryptocurrencies and financial markets and reported limited integration between crypto and traditional assets during Bitcoin's early growth phase.<sup><a href="#ref2">2</a></sup> These findings substantiate the early narrative that Bitcoin could serve as a portfolio diversifier due to its independence from typical macro markets.

However, more recent studies suggest that the Covid-19 pandemic marked a structural shift in asset correlations. Conlon and McGee examined Bitcoin’s performance during the March 2020 market crash and found it failed to act as a safe haven. Instead, Bitcoin exhibited increasing co-movement in price with equites during extreme volatility.<sup><a href="#ref3">3</a></sup> In a broader time-varying analysis, Dutta et al. documented dynamic correlations between gold, and stock markets, finding that Bitcoin’s relationship with equities strengthened during periods of market stress.<sup><a href="#ref4">4</a></sup> These findings suggest that correlations involving Bitcoin are not static but depend heavily on macroeconomic regimes.

Despite this growing body of literature, relatively few student or open-source data science projects explicitly examine how correlations between Bitcoin and multiple macro assets evolve across distinct time regimes, such as pre-Covid, pandemic, and post-pandemic periods; at least that we were able to find. Our project builds on existing literature by performing a regime-based correlation analysis across several major asset classes–the S&P 500, gold, and the U.S. dollar index–over the period 2017-2025. By combining rolling correlations and period segmentation, this work aims to evaluate whether Bitcoin has structurally transitioned from an uncorrelated alternative currency into a macro-sensitive risk asset.

## Hypothesis


From 2017 to 2025 in the United States, the relationship between price returns of Bitcoin and traditional macro-financial assets—the S&P 500 Index, gold price, and the U.S. Dollar Index (DXY)—changed significantly across the pre-COVID (2017–2019), during-COVID (2020–2022), and post-COVID (2023–2025) periods. Prior to the COVID-19 pandemic, Bitcoin exhibited weak or negligible correlations with equities, gold, and the U.S. dollar, consistent with its characterization as an alternative or speculative asset. During the COVID-19 period, Bitcoin’s correlation with the S&P 500 increased significantly while its correlations with gold and the dollar failed to strengthen in a manner consistent with safe-haven behavior, reflecting heightened integration with risk-on financial markets (i.e. stocks) during periods of systemic stress. In the post-COVID period, these elevated equity correlations persisted or stabilized, suggesting that Bitcoin has structurally transitioned toward behaving as a macro-sensitive risk asset rather than an independent currency. This hypothesis is tested by examining changes in correlation structure and regression coefficients across the three time regimes using daily data.

### Bitcoin Daily Closing Prices

This dataset contains daily closing prices of Bitcoin (BTC), one of the most widely traded cryptocurrencies. The primary metric in this dataset is:
- Closing Price (`Close`): The final trading price of Bitcoin on a given day, measured in USD. This value reflects the market consensus on Bitcoin's worth at the end of a trading day. Unlike traditional stock markets, cryptocurrency markets are open 24/7, so "daily close" usually refers to the price at 00:00 UTC or another standardized cutoff time.

Other columns that appear in the dataset include:
- Date (`date`): The calender day corresponding to each price entry
- Open, High, Low (`USD`): Optional columns showing the opening, highest, and lowest price for the day; reflecting Bitcoin's daily volitility
- Volume: Another optional column showing the amount, in USD, of 'volume' of Bitcoin units traded that day.

Understanding this data is important for investors, analysts, and researchers interested in price trends, volotility, and correlations with other financial metrics. For instance, extreme fluctuatoins in price can indicate market instability, while trends over time can inform investor stategies and their portfolio diversification.

```python
# Method used to tidy data
def find_missing_dates(df):
    df['Date'] = pd.to_datetime(df['Date']) 
    # Sort just in case
    df = df.sort_values('Date')
    # Create full expected daily range
    full_range = pd.date_range(
        start=df['Date'].min(),
        end=df['Date'].max(),
        freq='D'
    )
    return full_range.difference(df['Date'])

## YOUR CODE TO LOAD/CLEAN/TIDY/WRANGLE THE DATA GOES HERE
df_btc = pd.read_csv('./data/00-raw/btc-usd.csv')

df_btc = df_btc.drop([0, 1]) # csv file previously had multi-index columns. The column
                             # names are now entries. This removes them.

df_btc = df_btc.rename(columns={ # Rename columns that are useful to us
    'Price': 'Date',
    'Close': 'BTC_USD Close'
}).reset_index(drop=True)

#df_btc = df_btc.drop([13, 14]) # Used to test find_missing_dates function

missing = find_missing_dates(df_btc) # Test if any dates are missing
print(missing)

print(df_btc.columns) # Used to prove that each column has its own variable

df_btc['Date'].is_unique # Used to prove no repeated dates

df_btc.info() # Used to prove that each value is a single cell

df_btc.describe(include='all') #

print(df_btc)
```

Fluctuations in “close” indicates changes in stock market performance relative to the previous day. A rise from 5,000 to 5,050 would indicate a 1% increase in the combined market value. We note that these thresholds are not deterministic reflections of the market.

### US Dollar Index Prices

Within this dataset there are six columns. They are date, close, high, low, open, and volume. The most important metric that will be used for the project is the close column. The Close column represents the final reported value of the U.S. Dollar Index at the end of each trading session. It is measured in index points. These are unitless values that represent the relative strength of the U.S dollar compared to a basket of major foreign currencies. Those being: the Euro (EUR), Japanese Yen (JPY), British Pound (GBP), Canadian Dollar (CAD), Swedish Krona (SEK), and Swiss Franc (CHF).

The close is the most important daily metric because it serves as the official benchmark for performance measurement. Daily returns are calculated using changes in closing prices. Technical indicators such as the Relative Strength Index (RSI) are also calculated using this metric, which is a tool used to measure the speed and change of price movements. Moreover, the close tells us whether the dollar strengthend or weakened in relative to a previous day. For example, if on some day the dollar closes at 80 and the next day the dollar closes at 90, this means the dollar has strengthened, whereas if the dollar first closed at 90 and then closed at 80 this would mean the dollar has weakened.

For the past few decades, the USD index has fluctuated between 80 and 115. Values above 110 generally reflect a strong dollar, whereas values below 85 suggest a weak dollar.

In short, the close column for the USD Index lets us measure the relative strength and weakness of the US Dollar over a period of time.

There are no major concerns with the data. The data was taken from Yahoo Finance, which is a trusted source for market statistics.
