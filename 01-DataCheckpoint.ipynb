```python
# Method used to tidy data
def find_missing_dates(df):
    df['Date'] = pd.to_datetime(df['Date']) 
    # Sort just in case
    df = df.sort_values('Date')
    # Create full expected daily range
    full_range = pd.date_range(
        start=df['Date'].min(),
        end=df['Date'].max(),
        freq='D'
    )
    return full_range.difference(df['Date'])
```

```python
## YOUR CODE TO LOAD/CLEAN/TIDY/WRANGLE THE DATA GOES HERE
df_btc = pd.read_csv('./data/00-raw/btc-usd.csv')

df_btc = df_btc.drop([0, 1]) # csv file previously had multi-index columns. The column
                             # names are now entries. This removes them.

df_btc = df_btc.rename(columns={ # Rename columns that are useful to us
    'Price': 'Date',
    'Close': 'BTC_USD Close'
}).reset_index(drop=True)

#df_btc = df_btc.drop([13, 14]) # Used to test find_missing_dates function

missing = find_missing_dates(df_btc) # Test if any dates are missing
print(missing)

print(df_btc.columns) # Used to prove that each column has its own variable

df_btc['BTC_USD Close'] = pd.to_numeric(df_btc['BTC_USD Close'], errors='coerce') # Convert obj to numeric type
nan_count = df_btc['BTC_USD Close'].isnull().sum() # Count NaNs in column
#print(nan_count)

df_btc['Date'].is_unique # Used to prove no repeated dates

df_btc.info() # Used to prove that each value is a single cell

df_btc.describe(include='all') # Basic stats

print(df_btc)
```

### US Dollar Index
Within this dataset there are six columns. They are date, close, high, low, open, and volume. The most important metric that will be used for the project is the close column. The Close column represents the final reported value of the U.S. Dollar Index at the end of each trading session. It is measured in index points. These are unitless values that represent the relative strength of the U.S dollar compared to a basket of major foreign currencies. Those being: the Euro (EUR), Japanese Yen (JPY), British Pound (GBP), Canadian Dollar (CAD), Swedish Krona (SEK), and Swiss Franc (CHF).

The close is the most important daily metric because it serves as the official benchmark for performance measurement. Daily returns are calculated using changes in closing prices. Technical indicators such as the Relative Strength Index (RSI) are also calculated using this metric, which is a tool used to measure the speed and change of price movements. Moreover, the close tells us whether the dollar strengthend or weakened in relative to a previous day. For example, if on some day the dollar closes at 80 and the next day the dollar closes at 90, this means the dollar has strengthened, whereas if the dollar first closed at 90 and then closed at 80 this would mean the dollar has weakened. 

For the past few decades, the USD index has fluctuated between 80 and 115. Values above 110 generally reflect a strong dollar, whereas values below 85 suggest a weak dollar.

In short, the close column for the USD Index lets us measure the relative strength and weakness of the US Dollar over a period of time.

There are no major concerns with the data. The data was taken from Yahoo Finance, which is a trusted source for market statistics.
