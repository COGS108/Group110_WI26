## Research Question

Did Bitcoin behave more like a safe-haven asset or risk asset before, during, and after the COVID-19 pandemic (2017-2025), based on its correlations with the S&P 500, Gold price index, and U.S. Dollar index?

## Background and Prior Work

Over the past decade, cryptocurrencies–especially Bitcoin—have transitioned from niche digital assets into instruments increasingly integrated into global financial markets. As governmental and institutional participation has grown, researchers and investors have become more interested in understanding how Bitcoin interacts and compares with traditional macro assets. Namely: equities, commodities and currencies are subject to being juxtaposed with Bitcoin in consideration of their comparative investment advantages during such an unbridled economic period such as the Covid-19 era. 
We define the “Covid-19 era” to take place from the year 2017 through 2025, for our purposes. Specifically, the “pre-Covid era” will be the three years before the pandemic’s inception: 2017-2019; the “Covid era” will be the three years following its inception including its peak: 2020-2022; and the “post-Covid era” will be the three years following its peak: 2023-2025. Understanding whether Bitcoin behaves more like a speculative technology stock, a safe haven such as gold, or a liquidity sensitive macro asset is important for portfolio construction, risk management, and macroeconomic analysis.

Prior research suggests Bitcoin price historically exhibited low correlations with traditional assets before 2020. For example, Baur et al., analyzed Bitcoin's statistical properties and role as a diversification asset and found weak correlations with stocks and gold, concluding that bitcoin behaved more like a speculative investment than a safe haven or hedging tool.<sup><a href="#ref1">1</a></sup> Similarly, Corbet et al. examined spillovers between cryptocurrencies and financial markets and reported limited integration between crypto and traditional assets during Bitcoin's early growth phase.<sup><a href="#ref2">2</a></sup> These findings substantiate the early narrative that Bitcoin could serve as a portfolio diversifier due to its independence from typical macro markets.

However, more recent studies suggest that the Covid-19 pandemic marked a structural shift in asset correlations. Conlon and McGee examined Bitcoin’s performance during the March 2020 market crash and found it failed to act as a safe haven. Instead, Bitcoin exhibited increasing co-movement in price with equites during extreme volatility.<sup><a href="#ref3">3</a></sup> In a broader time-varying analysis, Dutta et al. documented dynamic correlations between gold, and stock markets, finding that Bitcoin’s relationship with equities strengthened during periods of market stress.<sup><a href="#ref4">4</a></sup> These findings suggest that correlations involving Bitcoin are not static but depend heavily on macroeconomic regimes.

Despite this growing body of literature, relatively few student or open-source data science projects explicitly examine how correlations between Bitcoin and multiple macro assets evolve across distinct time regimes, such as pre-Covid, Covid, and post-Covid periods; at least that we were able to find. Our project builds on existing literature by performing a regime-based correlation analysis across several major asset classes–the S&P 500 (stocks), gold (commodity), and the U.S. dollar index (currency)–over the years of 2017-2025. By combining rolling correlations and period segmentation, this work aims to evaluate whether Bitcoin has structurally transitioned from an uncorrelated, technological alternative currency into a macro-sensitive risk asset.

### US Dollar Index Prices

Within this dataset there are six columns. They are date, close, high, low, open, and volume. The most important metric that will be used for the project is the close column. The Close column represents the final reported value of the U.S. Dollar Index at the end of each trading session. It is measured in index points. These are unitless values that represent the relative strength of the U.S dollar compared to a basket of major foreign currencies. Those being: the Euro (EUR), Japanese Yen (JPY), British Pound (GBP), Canadian Dollar (CAD), Swedish Krona (SEK), and Swiss Franc (CHF).

The close is the most important daily metric because it serves as the official benchmark for performance measurement. Daily returns are calculated using changes in closing prices. Technical indicators such as the Relative Strength Index (RSI) are also calculated using this metric, which is a tool used to measure the speed and change of price movements. Moreover, the close tells us whether the dollar strengthend or weakened in relative to a previous day. For example, if on some day the dollar closes at 80 and the next day the dollar closes at 90, this means the dollar has strengthened, whereas if the dollar first closed at 90 and then closed at 80 this would mean the dollar has weakened.

For the past few decades, the USD index has fluctuated between 80 and 115. Values above 110 generally reflect a strong dollar, whereas values below 85 suggest a weak dollar.

### Stationarity Tests

- Stationarity is a concept in **time series analysis**. A time series is *stationary* if its statistical properties stay constant over time.
In simple terms:
A stationary series behaves the same way throughout the entire timeline.
- In order to learn if there exists any correlations between the Bitcoin Returns and the returns of other variables, we must first test each financial variable for each era for *stationarity.* If they all are, we can proceed to conduct rolling correlations.

```python
# Let's define a helper function to help us test each finacial class variable's stationarity across 
# all three time regimes to be confident that rolling correlations are appropriate

from statsmodels.tsa.stattools import adfuller

def run_adf_test(series, series_name="Series"):
    """
    Runs Augmented Dickey-Fuller test and prints formatted output.
    """
    result = adfuller(series.dropna())
    
    print(f"\nADF Test Results for {series_name}")
    print("-" * 40)
    print(f"p-value: {result[1]}")    
    if result[1] < 0.05:
        print("Conclusion: Reject H0 → Series is stationary.")
    else:
        print("Conclusion: Fail to reject H0 → Series is non-stationary.")
```
